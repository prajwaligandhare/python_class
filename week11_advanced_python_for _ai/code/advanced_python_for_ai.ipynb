👨‍🏫 Concept 2 — generator (yield) — ek-ek karke do

```python
def first_n_sqaures(n):
    for i in range(n):
        yield i ** 2

squares = first_n_sqaures(5)
print(squares)

for sq in squares:
    print(sq)
```

💻 Demo — fake token streamer (LLM jaisा!)

```python
import time

def stream_response(text):
    """LLM jaisा! - ek-ek word 'stream' karo."""
    for word in text.split():
        yield word
        time.sleep(0.1)

for token in stream_response("Hello I am an AI agent, How can I help you today?"):
    print(token, end=" ", flush=True)        
```

```python
def my_decorator(func):
    def wrapper():
        print("Before the function")
        func()
        print("After the function")
    return wrapper

def say_hello():
    print("Hello")  

decorated = my_decorator(say_hello)
decorated()          
```

👨‍🏫 Concept 3 — @ syntax (sundar shortcut)

```python
def my_decorator(func):
    def wrapper():
        print("Before the function")
        func()
        print("After the function")
    return wrapper 

@my_decorator
def say_hello():  
    print("Hello")

say_hello()     
```

👨‍🏫 Concept 4 — arguments wale functions ko wrap karna (*args, **kwargs)

```python
import functools

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        import time
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds to run")
        return result
    return wrapper

@timer
def slow_add(a, b):
    import time
    time.sleep(0.5)
    return a + b

print(slow_add(2, 3))            
```

##CLASS - Context Managers & Collections

Concept 1 - context manager kya karta hai

```python
from contextlib import contextmanager
import time
```

```python
@contextmanager
def timer(label):
    start_time = time.time()
    print(f"Starting {label}...")
    print("Starting time is", start_time)
    yield
    end_time = time.time()
    print(f"{label} took {end_time - start_time} seconds")
```

```python
with timer("my task"):
    total = sum(range(1000))
    print("Total Sum is", total)
```

Concept 2 - dataclass with defaults & methods

```python
from dataclasses import dataclass, field
```

```python
@dataclass
class Agent:
    name: str
    model: str = "gpt.4o"
    tools: list = field(default_factory = list)


    def add_tool(self, tool):
        self.tools.append(tool)  
```

```python
agent_a  = Agent("agent_a", tools=["tool_1", "tool_2"])
agent_2 = Agent("agent_2", model="gpt.4o", tools=["tool_3", "tool_4"])
```

```python
@lru_cache(maxsize=None)
def slow_square(n):
    print(f"Calculating {n} square.....")
    return n ** 2
```
