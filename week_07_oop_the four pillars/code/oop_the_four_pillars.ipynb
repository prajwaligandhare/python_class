CLASS 43 - Inheritance

Concept 1 - basic Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

    def sleep(self):
        print(f"{self.name} is sleeping")   

cow = Animal("Cow")  
cow.eat()      
```

```python
class Dog(Animal):
    def bark(self):
        print(f"{self.name} is barking")

dog1 = Dog("Tommy")   
dog1.bark()  
dog1.eat()
dog1.sleep()   




```

Concept 3 — super() (parent ka setup use karo)

```python
class Dog(Animal):
    def __init__(self, name, breed_name):
        super().__init__(name)
        self.breed_name = breed_name

    def bark(self):
        print(f"{self.name} is barking")

    def breed(self):
        print(f'{self.name} is {self.breed_name}')

dog2 = Dog("Tiku", 'German Sheherd')


```

```python
dog2.bark()
```

```python
dog2.breed()
```

```python
dog2.sleep()
```
