1. Food Delivery

A customer orders food worth ₹850.

If the order is above ₹1000 → 20% discount.
Else if the order is above ₹500 → 10% discount.
Otherwise → No discount.

Print:

Original amount
Discount
Final amount

```python
cust_ord_price = int(input("Enter the customer order's price:  "))

discount = int(input("enter the discount: "))

final_amnt = 850 / cust_ord_price * 100

print(f"original amount is {cust_ord_price} & got {discount}%, final amount is {final_amnt} ")

if cust_ord_price >= 1000:
   print("Got 20% discount")

elif cust_ord_price >= 500:
     print("Got 10% discount")

else:
     print("No Discount")
     
             
```

2. Movie Ticket

Take:

Age
Ticket price

Rules:

Children (below 12) → 50% discount
Senior citizens (60+) → 30% discount
Others → No discount

Print the final ticket price.

```python
age = int(input("Enter your age: "))

ticket_price = int(input("Enter the ticket price: "))

#discount = int(input("enter the discount: "))

if age <= 12:
   print("You are a children, you got 50% discount")

elif age >= 60:
     print("You are senior citizen, you got 30% discount")

else:
   print("No Discount")


final_price = ticket_price * 0.5

print(f"Final Price is : {final_price}")
```
