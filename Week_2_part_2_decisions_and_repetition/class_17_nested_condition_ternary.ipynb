## Concept 1 - Nested if (if ke andar if)

```python
## Concept 1 - Nested if (if ke andar if)

age = 20
has_ticket = True

if age >= 18:
    if has_ticket:
        print("enjoy the movie!")

    else:
        print("Please buy a ticket")
    
else:        
    print("only 18+ allowed")
```


## Concept 2 - Aksar and zyada saaf hota hai

```python
age = 20
has_ticket = True

if age >=18 and has_ticket:
    print("Enjoy the movie!")

else:

    print("Sorry, cannot enter")
        
```

##Concept 3 — Ternary (one-line if/else)


```python
age = 20
status = "Adult" if age >=18 else "Minor"
print(status)
```

## Demo - ticket price y age


```python
age = int(input("Enter your age: "))

if age < 5:
    price = 0
elif age < 18:
    price = 100
elif age < 60:
    price = 200
else:
    price = 50

print(f"Ticket price: {price}")
            
```
