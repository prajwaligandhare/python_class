Concept 1 - for loop ek list par


```python
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print(fruit) 
```

 Concept 2 — range() — numbers banata hai

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 10):
    print(i, end= " ")
```

Concept 3 — enumerate (number + item dono)

```python
fruits = ["apple", "banana", "mango"]

for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

Concept 4 — zip (do lists saath-saath)

```python
names = ["Asha", "Rahul", "Raj"]
marks = [90, 85, 95]

for name, mark in zip(names, marks):
    print(f"{name}: {mark}")
```

Concept 5 — reversed (ulta chalo)

```python
for i in reversed(range(1, 6)):
    print(i)
```

```python
# Demo — multiplication table

num = int(input("Which table? "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
    
```

```python
items  = ['laptop', 'mouse', 'keyboard', 'monitor']

for item in items:
    if item == 'laptop':
        print("I bought 3 laptops")

    elif item  == 'mouse':
        print("I bought 4 mouse")  

    elif item  == 'keyboard':
        print("I bought 2 keyboards")

    elif item == 'monitor':
        print("I bought 1 monitor")    

    else:
        print("Invalid items")        
```

```python
# ZIP


items  = ['laptop', 'mouse', 'keyboard', 'monitor']

items_count = [4, 3, 2, 1]

for item_count, item in zip(items_count, items):
    print(f"bought {item_count} {item}")
```

```python
shops_nm = ['shop1', 'shop2', 'shop3', 'shop4']

fruits = ['banana', 'apple', 'orange', 'muskmelon']

counts_kg = ['10kg', '15kg', '20kg', '25kg']

for shop_nm, fruit, count_kg in zip(shops_nm, fruits, counts_kg):
    print(f"from {shop_nm} bought {count_kg} {fruit}")



```

#Enumerate

```python
fruits = ['banana', 'mango', 'apple', 'orange']

for index, fruit in enumerate(fruits):
    print(f"bought {index} {fruit}")
    print("-"*20)
    
```

```python
emps_id = [101, 102, 103, 104]

for emp_id in enumerate(emps_id):
    print(f"employee {emp_id} got grades")

    print("*" * 30)
    
    print("Congratulation All !....")

    print("-"*30)
```

#While loop

```python
you_are_not_full = True

while you_are_not_full:
    user_input = input("If you feal full enter Y else N ?")

    if user_input == 'Y':
        you_are_not_full = False
    else:
        you_are_not_full = True    

```
