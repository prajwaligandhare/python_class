concept 2- match/Case (saaf routing)

```python
user_input = input("Enter command for s/w (start/stop/paus): ")

match user_input:
    case "start":
         print("s/w starting.....")

    case "stop":
         print("s/w stopping.....")

    case "paus":
        print("s/w paused")

    case _:
         
          print("Unknown coomand")


        
                 
```
