Menu calculator ko run karke saare 4 operations test karo.


```python
print("==============Calculator==============")

while True:

    num1 = int(input("Enter the first number:  "))

    num2 = int(input("Enter the second number: "))

    operators = input("Enter the operators (+, -, *, /): ")

    if operators == "+":
        print(f"The adition of {num1} {operators} {num2} is: ", num1+num2)

    elif operators == "-":
        print(f"The substraction of {num1} {operators} {num2} is: ", num1-num2)

    elif operators == "*":
        print(f"The multiplication of {num1} {operators} {num2} is: ", num1*num2)
    

    elif operators == "/":
        print(f"The division of {num1} {operators} {num2} is: ", num1/num2)


    else:
        print("Invalid operator")   


    choice = input("Do you want to continue ? (yes/no)")

    if choice == "no":
        break

print("Thank you for using this calculator.......!")



   
```

Number guessing game khelo aur jeeto!

```python
import random

print("===========Welcome To Guessing Game================")

secret_number = random.randint(0, 100)

while True:

     secret_number = random.randint(0, 100)

     print("To Win you hav only 4 Ateempts")

     for i in range(0, 4):
          print(f"Attempts {i+1} of 4")
          guss_nm = int(input("Enter the guess number [0 - 100]: "))

          if guss_nm == secret_number:
           print("Congratullations..........!")
           break

          elif guss_nm > secret_number:
               print("Too high.....!")

          elif guss_nm < secret_number:
               print("Too low......!")     

          else:
           print("Better Luck Next Time....!")

     print(f"The secret number was: {secret_number}")

     print("="*50)

     choice = input("Do you want to continue ? (yes/no)")

     if choice == "no":
          break

print("Thank you for playing this game")
    
       
```
