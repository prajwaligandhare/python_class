Discount Calculator

Input the original price of a product.
Input the discount percentage.
Calculate the discount amount.
Print the final price.

```python

product_nm = input("Enter the product name: ")
original_price = int(input("Enter the product price: "))

discount_per = int(input("Enter the discount percentage %: "))

discount_am = original_price * (100 - discount_per)/100

print(f" The product price is {original_price} rupees and we got {discount_per}% discount on {product_nm} and final price is: {discount_am} rupees")


```

GST Calculator

Input the price of an item.
Input the GST percentage.
Calculate the GST amount.
Print the final amount after adding GST.

```python
item = int(input("Enter The price of item: "))

GST_per = int(input("Enter The GST Percentage: "))

GST_amnt = item * GST_per / 100


final_price = item + GST_amnt

print(f"The final price after adding GST {GST_per}% is: {final_price}%")
```

Salary Increment

Input an employee's current salary.
Input the increment percentage.
Calculate the new salary after the increment.

```python
current_salary = int(input("Enter the current salary:  "))

increment_per = int(input("Enter the increment percentage: "))

new_salary = current_salary * increment_per / 100

final_salary = new_salary + current_salary

print(f" The salary after the increment is: {final_salary}")


```
