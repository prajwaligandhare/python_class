CLASS 37 - Recursion

Concept 2 - Factorial

```python
def factorial(n):
    if n == 1:
        return 1

    return n * factorial(n-1)  # 5, 4, 3, 2, 1 = 120
```

```python
factorial(5)
```

CLASS 39 - lambda, map & filter

Concept 1 - lambda (mini one-line function)

```python
def square(x: int) ->int:
    return x ** 2
```

```python
square(2)
```

```python
# correct way using lambda

square_llm = lambda x: x**2
square_llm(3)
```

```python
even_odd_num = lambda n: "Even value" if n % 2 == 0 else "Odd value"
```

```python
even_odd_num(12)
```

```python
check_neg_pos_num = lambda num: 'negative' if num < 0 else 'positive'
check_neg_pos_num(1)
```

Concept 2 - map() (har item par function lagao)

map(function, item)

```python
nums = [1, 2, 3, 4, 5]
```

```python
sqr_num = list(map(lambda n: n ** 2, nums))
sqr_num
```

```python
num_1= [-1, 3, -2, 5, -7]

t_f_value = list(map(lambda n: False if n < 0 else True, num_1))

t_f_value
```
