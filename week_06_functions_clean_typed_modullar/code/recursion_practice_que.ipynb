1. Print numbers from 1 to n

```python
def numbers(n): # 3, 2, 1, 0
    if n == 0:  #3.then jump to inside numbers(3).
                # check condition 3 == 0 >> False.
                # 2 == 0 >> False.
                # 1 == 0 >> False.
                # 0 == 0 >> True.

       return  # 4. So python skip the return & jump to next line.
               # if condition is true then return the value.
               # The function immediately ends, Nothing is printed.

    numbers(n-1) #5. Now python jump to numbers(3-1) = 2.
                 # it dose not execute next line (print(n)).
                 # it will return to previous function call numbers(2).
                 # 2 -1 = 1.
                 # 1 -1 =0.
    print(n) 

numbers(3) #1. call stack >> first function call.
           #2. current value n = 3   
           # call stack becomes 3, 2,1  
    
```

```python
#print the values from 1 to n:

def sum_function(n): # 5, 4, 3, 2, 1, 0
    if n == 0:  # 5 == 0 >> false
                # 4 == 0 >> false
                # 3 == 0 >> false
                # 2 == 0 >> false
                # 1 == 0 >> false
                # 0 == 0 >> true
       return print("Return value is: ", n)

    sum_function(n-1) # 5 -1 = 4.
                      # 4 -1 = 3.
                      # 3 -1 = 2.
                      # 2 -1 = 1.
                      # 1 -1 = 0 >> return the value, not consider while printing.
    print(n) 

sum_function(5)      
```

2. Print numbers from n to 1

```python
def revers_num(n): #5,4,3,2,1,0
    if n == 0: # 5 == 0 >> f
               # 4 ==0 >> f, 3==0>>f, 2==0>>f,1==0>>f
               # 0==0>> T
       return #>> return 0 >> end the loop with 0.
    print(n) # 5, 4, 3,2,1

    revers_num(n-1) # 5-1=4, 4-1=3, 3-1=2, 2-1=1, 1-1=0

revers_num(5)       
```

3. Find factorial

```python
def factorial(n): #5,4,3,2,1,0
    if n == 0: # 5==0 >> f, 4==0>>f, 3==0>>f, 2==0>>f, 1==0>>f
               # 0==0 >>T
       return # return 0.

    return n * factorial(n-1) # 5-1=4, 4-1=3, 3-1=2, 2-1=1, 1-1=0                                    
                                     
 # 5*4*3*2*1

print(factorial(5))      
```
