👨‍🏫 Concept 2 — set operations (maths jaise)

📖 Technical definition — Set operations: Set operations combine or compare sets:
 union (|) returns all elements from both sets, intersection (&) returns elements common to both, 
 and difference (-) returns elements in the first set that are not in the second.

```python
a = {1, 2, 3, 4}

b = {3, 4, 5, 6}

print(a | b)
print(a & b)
print(a - b)
```
