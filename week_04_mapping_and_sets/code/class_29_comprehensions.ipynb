Concept -2 condition ke sath (filter)

```python
even_number = [num for num in range(1, 11) if num % 2 == 0 ]
even_number
```

```python
odd_number = [n for n in range(1, 11) if n % 2 != 0]
odd_number
```

```python
#old technique:

square = []

for n in range(1, 6):
    square.append(n**2)
print(square)    
```

```python
#Ab wahi kaam, ek line mein — comprehension:"

square = [n**2 for n in range(1, 6)]
square
```

 Concept 4 — dict aur set comprehension

```python
#dict comprehension

nums = [1, 2, 3, 4]

square = {n: n ** 2 for n in nums}

print(square)
```

```python
 #do lists ko dict banao (zip yaad hai?)
names = ['mohan', 'kumar']
marks = [85, 79]

students_mrk = {name: mark for name, mark in zip(names, marks)}
print(students_mrk)
 

```

```python
# set comprehension

words = ['Apple', 'BANANA', 'mango']

unique_len = {len(w) for w in words}

print(unique_len)


```

```python
uniq_length = {len(w) for w in ['hi','bye', 'ok']}
print(uniq_length)
```

```python
#Demo — clean a list of user inputs

raw = ["  asha", "RAHUL", 'priya  ', ""]

# strip + lowercase + khaali hatao, sab ek line mein

cleaned = [name.strip().lower() for name in raw if name.strip()]

print(cleaned)
```
