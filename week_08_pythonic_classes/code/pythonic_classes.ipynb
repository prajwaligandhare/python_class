Concept 2 - __eq__ (apni equality define karo)

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y    


p1 = Point(1, 2)
p2 = Point(1, 2)    

p1 == p2
```

CLASS 48 - classmethod & staticmethod

Concept 1 - @classmethod (alternative constructor)

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, data):
        name, age = data.split(",")
        return cls(name, int(age))
```

```python
p1 = Person('ayush', 23)
```

```python
p1.name, p1.age
```

```python
p2 = Person.from_string("ayush, 23")
```

```python
p2.name, p2.age
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_dict(cls, data):
        name, age = data["name"], data["age"]
        return cls(name, age)
```

CLASS 49 - Property

Concept 1 - Computed property (apne aap calculate)

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14 * self.radius ** 2    
```

```python
circle = Circle(5)

circle.area
```

```python
circle.radius = 10
circle.area
```

```python
class Temperature:
    def __init__(self, celcius = 0):
        self._celcius = celcius

    @property
    def celcius(self):
        return self._celcius   


    @celcius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Below absolute zero!")    

        self.celcius = value    
```

```python
temp = Temperature()
```

```python
temp.celcius
```

```python
temp.celcius = 6
```

```python
temp.celcius
```

```python
print(temp.celcius)
```

```python
temp.celcius = -275
```

```python
class Temperature:
    def __init__(self, celsius=0):
        self._celsius = celsius

    @property
    def celsius(self):              # getter — value padhne par
        return self._celsius

    @celsius.setter
    def celsius(self, value):       # setter — value set karne par check
        if value < -273.15:
            raise ValueError("Below absolute zero!")
        self._celsius = value
```

t = Temperature()

Concept 1 — composition = object ke andar object


```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"{self.horsepower} HP Engine Started"  

class Car:
    def __init__(self, brand, horsepower):
        self.brand = brand
        self.horsepower = horsepower
        self.engine = Engine(self.horsepower).start()

    def start(self):
        return self.engine     

class Car:
    def __init__(self, brand, horsepower):
        self.brand = brand
        self.horsepower = horsepower
        self.engine = Engine(self.horsepower).start()
        self.colour = self.colour()    

    class Features:
        def __init__(self, colour):
            self.colour = colour

    def start(self):
        return self.engine

    def colour(self, colour = "red"):
        return self.Features(colour).colour                        

     


     
```
