##👨‍🏫 Concept 1 — items add karna


📖 Technical definition — List method: A list method is a built-in operation attached to a list object that inspects 
or modifies the list, such as append(), insert(), remove(), pop(), extend(), sort(), and reverse(). 
Because lists are mutable, many of these methods change the list in place.
