Concept 4 — slicing (ek hissa kaatna)

```python
num = [10, 20, 30, 40, 50, 60, 70] 
       #0  1   2   3   4    5   6

print(num[0:3])
print(num[:5])
print(num[1:])

print(num[0:5:2])

print(num[::3])

print(num[0:4:2])

print(num[::-1])

print(num[-5:-1])

print(num[-6:-3])



```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print(f"First day: {days[0]}")

print(f"staring week day: {days[-7]}")

print(f"Second day: {days[-6]}")

print(f"week's second day: {days[1]}")

print(f"Weekdays: {days[:5]}")

print(f"Weekends: {days[-2:]}")
```
