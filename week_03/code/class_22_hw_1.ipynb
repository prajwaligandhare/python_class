```python
#1.Khaali list mein 5 numbers append karo, phir max, min, aur average print karo.



my_list = []
my_list.append(20)
my_list.extend([23, 50, 70])
my_list.append(80)

maximum =max(my_list)
print(f"Maximum number is {maximum}")

minimum = min(my_list)
print(f"Minimum number is {minimum}")


average = sum(my_list) / len(my_list)
print(f"Average number is {average}")


my_list


```

```python
#2.Ek list ko sorted() se sort karo BINA original badle (dono print karke dikhao).

city = ["Mumbai", "Delhi", "Bangalore", "Hyderabad", "Chennai"]

sorted_city = sorted(city)

print(city)

print(sorted_city)




```

```python
#3.Ek list se pop() se aakhri item nikaalo aur use print karo.

city = ["Mumbai", "Delhi", "Bangalore", "Hyderabad", "Chennai"]

last_item = city.pop(-1)

print(last_item)



```
