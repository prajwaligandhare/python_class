```python
#3. Ek 3x3 grid ke saare items nested loop se print karo.

grid = [
        [12, 23, 10],
        [45, 33, 9],
        [5, 44, 50]
]

for row in grid:
    for col in row:
       print(col, end = " ")
    print()
```
