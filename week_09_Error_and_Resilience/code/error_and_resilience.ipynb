👨‍🏫 Concept 3 — else aur finally

```python
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("That's not a valid number!")
else:
    print(f"Great! You entered {number}")
finally:
    print("Thank you for using the program")            
```

```python
try:
    number = int(input("Enter a number: "))
except ValueError as e:
    print(f"Error: {e}")
else:
    print(f"Well done, you have entered a valid number: {number}")
finally:
    print("this block always runs")        
```

try — risky code.

except — error aaye toh.

else — koi error NA aaye toh (success).

finally — HAMESHA (error ho ya na ho) — cleanup ke liye.

👨‍🏫 Concept 3 — custom exception class

```python
class InsufficientFundsError(Exception):
    """Raised when withdrawl exceeds the balance."""
    pass
```

```python
def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Cannot withdraw {amount}, balance is {balance}")
    return balance - amount  

try:
    withdraw(100, 500)
except InsufficientFundsError as e:
    print(f"Transaction Failed: {e}")
```

```python
print(withdraw(10000, 500))
```

```python
print(withdraw(100, 500))
```
