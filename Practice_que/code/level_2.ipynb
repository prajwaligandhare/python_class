6. Login System

   Username = "admin"

   Password = "python123"
   
   Print Login Successful or Invalid Credentials.
   

```python
username ="admin"

password = "python"

user = input("Enter the username: ")
pword = input("Enter the password: ")

if user == username and pword == password:
     print(f"login succesfully: you enter correct username: {user} & password: {pword}")

else:
     print(f"Invalid Credential: you enter in-correct username: {user} & password: {pword}")    
      
```
