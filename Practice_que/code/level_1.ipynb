Take age as input.

If age < 18 → Child
18–59 → Adult
60+ → Senior Citizen

```python
age = int(input("Enter your age: "))

if age < 18:
   print("You are a child")

elif age >= 18 and age <=59:
     print("You are a adult")

elif age >=60:
   
    print("You are a senior citizen")  

else:
    print("Invalid age")
    
            
```

Take marks of 5 subjects.

Calculate total
Percentage
Print Pass if percentage ≥ 35 else Fail.

```python
math = int(input("Enter the math marks: "))

eng = int(input("Enter the english marks: "))

marathi = int(input("Enter the marathi marks: "))

hindi = int(input("Enter the Hindi marks: "))

hist = int(input("Enter the History marks: "))

total_marks = math+eng+marathi+hindi+hist

percentage = total_marks/500 * 100

if percentage >= 35:
   print(f"You are Pass with {percentage}%")

else:
   print(f"you are Fail with {percentage}%")
```
