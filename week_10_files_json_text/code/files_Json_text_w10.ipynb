Concept 2 - modes (w vs a vs r)

w - write (overwrite data)

a - append

r - Read

```python
with open("my_first_file.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
with open("my_first_file.txt", "a") as f:
    t = "\nHello, This is Appends things."
    f.write(t)
```

```python
with open("my_first_file.txt", "r", encoding="utf-8") as f:
    f.write("Hello, नमस्ते 🙏😊 positivity")
    
```

#### Small Project : Diary

```python
def add_entry(text: str)->None:
    with open("diary.txt", "a") as f:
        f.write(text + "\n")

def read_diary()->str:
    with open("diary.txt", "r") as f:
        for line in f:
            print(line.strip())
        return f.read()    


add_entry("I am learning Python with Indian AI Production at 9 AM")
add_entry("Lunch break at 12PM")            
```

```python
read_diary()
```

```python
def fruits(veg_txt: str)->None:
    with open("Fruits_file.txt", "a") as f:
        f.write(veg_txt + "\n")

def vegetable()->str:
    with open("Fruits_file.txt", "r") as f:
        for line in f:
            print(line.strip())
        return f.read()    

fruits("Apple") 
fruits("Chikkuu")         
```
