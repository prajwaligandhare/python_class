### CLASS 59 — CSV & Regex

👨‍🏫 Concept 1 — CSV (comma-separated values)

```python
import csv

rows = [
         ["Name", "Marks"],
         ["Asha",   85],
         ["Rahul",  90]
]

with open("marks.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

```python
with open("marks.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

Concept 2 - CSV as dictionaries (zyada readable)

```python
with open("marks.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row_dict in reader:
        print(row_dict["Name"], row_dict["Marks"])
```

Concept 3 - Regex (patterns dhoondhna)

```python
import re
```

```python
text = """My contact: email >> prajedsfdf@gmail.com,
          and my mobile no >> 12345677, my secondary mo no: 76532257 """

numbers = re.findall(r"\d+", text)
print(numbers)         
```

```python
numbers = re.search(r"\d+", text)
print(numbers.group())
```

```python
list_words = text.split(" ")
```

```python
list_words
```

```python
for word in list_words:
    if "@" in word:
        print(word)
```

```python
def get_email(text):
    list_words = text.split(" ")
    for word in list_words:
        if "@" in word:
            print(word)

get_email(text)            
```
